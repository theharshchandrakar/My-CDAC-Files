Draw line & bar charts to show urban and rural population of India over the years using population.csv [rural population is the difference between total population and urban population][For bar chart consider 8 yrs data]


```python
data=pd.read_csv("Population.csv")
```

```python
data
```

```python
data["Rural Population"]=data["Population"]-data["Urban Population"]
```

```python

plt.plot(data["Year"],data["Urban Population"],label="Urban population")
plt.plot(data["Year"],data["Rural Population"],label="Rural Population")
plt.xlabel="Year"
plt.ylabel="Urban Population"
plt.title="population"
plt.legend()

plt.show()
```
