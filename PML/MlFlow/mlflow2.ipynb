# ML Flow


```python
import mlflow

from sklearn.model_selection import KFold, GridSearchCV
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.pipeline import Pipeline
import pandas as pd
import numpy as np
import os
#os.chdir('G:/My Drive/1Practical Machine Learning/notes/Cases/Medical_Cost_Personal')
```

```python
mlflow.set_tracking_uri("http://localhost:5000")
```

```python
# Start an MLflow experiment (you can change the experiment name)
mlflow.set_experiment("Medical Insurance Charges - DTRegressor")
```

```python
mlflow.sklearn.autolog()
```

```python
med=pd.read_csv("D:\Machine Learning\Datasets\Cases\Medical_Cost_Personal\insurance.csv")
X, y = med.drop('charges', axis=1), med['charges']
```

```python
ohe = OneHotEncoder(drop='first', sparse_output=False).set_output(transform='pandas')
col_trnf = ColumnTransformer([('OHE', ohe, make_column_selector(dtype_include=object))], # type: ignore
                             remainder='passthrough',
                             verbose_feature_names_out=False)
col_trnf = col_trnf.set_output(transform='pandas')
X = col_trnf.fit_transform(X)
```

```python

from sklearn.tree import DecisionTreeRegressor
```

```python


dt = DecisionTreeRegressor()
pipe = Pipeline([('TRNS',col_trnf), ('DT', dt)])
# el.get_params()
params = {
    'DT__max_depth': [3, 5, 7, 10, 15, None],              # depth of the tree
    'DT__min_samples_split': [2, 5, 10, 20],               # min samples to split a node
    'DT__min_samples_leaf': [1, 2, 4, 10],                 # min samples in a leaf
    'DT__max_features': ['sqrt', 'log2', None],            # features considered at each split
    'DT__criterion': ['squared_error', 'absolute_error']   # splitting criterion (MSE or MAE)
}



kfold = KFold(n_splits=5, shuffle=True, random_state=25)

gcv = GridSearchCV(pipe, param_grid=params, cv=kfold, scoring='r2', verbose=3)
gcv.fit(X, y)
```

```python
print(gcv.best_params_)
print(gcv.best_score_)
```

```python
best=gcv.best_estimator_
```
