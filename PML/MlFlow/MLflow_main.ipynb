```python
import gradio as gr
with gr.Blocks() as demo:
    with gr.Row():
       age= gr.Number(label='Age')
       gender = gr.Dropdown(label="Gender", choices=["female", "male"], value="female")
       bmi = gr.Number(label='BMI')
    with gr.Row():
        child =gr.Number(label='Children')
        smoker = gr.Dropdown (label='Smoker', choices= ["yes", "no"], value="no")
        region= gr.Dropdown (label='region', choices= ["northeast","southwest","northwest","southeast"], 
                             value="southeast")
    with gr.Row():
        charges = gr.Number(label='Medical Charges')
    with gr.Row():
        button = gr.Button(value="Calculate")
        button.click(fn=predict, 
        inputs=[age, gender, bmi, child, smoker, region], 
        outputs=charges
    )

# --- LAUNCH THE APP ---
demo.launch()
```
