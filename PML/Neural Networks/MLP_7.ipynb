### pytorch

```python
import pandas as pd
import numpy as np 
import torch.nn as nn
import torch
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import MinMaxScaler, StandardScaler, LabelEncoder
```

```python
wisconsin = pd.read_csv('D:\Machine Learning\Datasets\Cases\Wisconsin\BreastCancer.csv')
le = LabelEncoder()
wisconsin['Class'] = le.fit_transform(wisconsin['Class'])
X,y = wisconsin.drop('Class', axis = 1), wisconsin['Class'].values
X_train , X_test , y_train , y_test = train_test_split(X,y,test_size=0.3,random_state=25,stratify=y)
scl_x = MinMaxScaler()
X_trn_scl,X_tst_scl = scl_x.fit_transform(X_train), scl_x.fit_transform(X_test)
```

```python
X_torch = torch.from_numpy(X_trn_scl)
y_torch = torch.from_numpy(y_train)
print(X_torch.size())
print(y_torch.size())
```

```python
torch.manual_seed(25)
model = nn.Sequential(nn.Linear(in_features=X_trn_scl.shape[1], out_features=5),
                      nn.ReLU(),
                      nn.Linear(5,3),
                      nn.ReLU(),
                      nn.Linear(3,1),
                      nn.Sigmoid())
criterion = torch.nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr = 0.001)
```

```python
for epoch in np.arange(0,1000):
    y_pred_prob = model(X_torch.float())
    loss = criterion(y_pred_prob[:,0], y_torch.float())
    print('epoch:', epoch+1, 'loss:', loss.item())
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
```

```python
from sklearn.metrics import log_loss


X_torch_test = torch.from_numpy(X_tst_scl)
y_pred_prob = model(X_torch_test.float())
y_pred_prob = y_pred_prob.detach().numpy()
y_pred = np.where(y_pred_prob >= 0.5, 1,0)
print(accuracy_score(y_test, y_pred))
print(log_loss(y_test,y_pred_prob))
```
