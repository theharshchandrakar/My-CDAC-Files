```python
import pandas as pd
import numpy as np
from chronos import Chronos2Pipeline
from sklearn.metrics import root_mean_squared_error
import os
os.chdir('D:\Machine Learning\Datasets')
import warnings
warnings.filterwarnings('ignore')
```

```python
pipline = Chronos2Pipeline.from_pretrained("amazon/chronos-2")
```

```python
df = pd.read_csv("monthly-milk-production-pounds-p.csv")
df['timestamp'] = pd.date_range(start='1962-01-01', periods=len(df), freq='M')
df['id'] = 0 # Should be identical
df.head()
```

```python
train = df.iloc[:-12]
test = df.iloc[-12:]
future_df = test.drop(columns=["Milk"]).copy()
future_df['id'] = 0
future_df['timestamp'] = pd.date_range(start='1962-01-01', periods=12, freq='ME')
```

```python
pred_df = pipline.predict_df(
    train, future_df=future_df,
    validate_inputs=False, #
    prediction_length=12,
    id_column="id",
    timestamp_column="timestamp",
    target='Milk',
)
pred_df.columns
y_pred = pred_df["predictions"]
```

```python
root_mean_squared_error(test['Milk'], y_pred)
```
