```python
from transformers import pipeline
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
data = {"Actors":['Brad Pitt', "Leo Di Caprio", "George Clooney"], "Number of Movies": ['87','53','69']}
table = pd.DataFrame.from_dict(data)
table
type(table)
```

```python
# Use a pipeline as a high-level helper
from transformers import pipeline

tqa = pipeline("table-question-answering", model="google/tapas-base-finetuned-wtq")
```

```python
question = "how many movies does brad pitt have ?"
print(tqa(table = table, query=question)['cells'][0])
```

```python
import gradio as gr
from io import StringIO
def answer(question, file):
    with open(file, 'r') as readfile:
        ctx = readfile.read().rstrip()
    csvStringIO = StringIO(ctx)
    df = pd.read_csv(csvStringIO, sep=",")
    df = df.astype('str')
    response = tqa(table=df, query= question)['cells'][0]
    return response
```

```python
demo = gr.Interface(
    fn = answer,
    inputs = ['text',gr.File(label="Information file")],
    outputs=["text"]
)
demo.launch()
```
