```python
os.chdir('D:\Machine Learning\Datasets')
df=pd.read_csv("D:\Machine Learning\Datasets\Cases\Glass_Identification\Glass.csv")
df.isna().sum()
```

```python
le=LabelEncoder()
df['Type']=le.fit_transform(df['Type'])
X,y=df.drop('Type',axis=1),df['Type']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25,stratify=y)
```

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train,y_train)
y_pred=knn.predict_proba(X_test)
```

```python
# Calculate log loss
logloss_without_scaler = log_loss(y_test, y_pred)
print(f"Log Loss (without scaler): {logloss_without_scaler:.4f}")
```

```python
# With StandardScaler
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)

knn_scaled=KNeighborsClassifier(n_neighbors=3)
knn_scaled.fit(X_train_scaled,y_train)
```

```python

y_pred_proba_scaled=knn_scaled.predict_proba(X_test_scaled)
logloss_with_scaler = log_loss(y_test, y_pred_proba_scaled)
print(f"Log Loss (with scaler): {logloss_with_scaler:.4f}")
```

```python
# Comparison
print("\n" + "=" * 50)
print("COMPARISON")
print("=" * 50)
print(f"Log Loss without scaler: {logloss_without_scaler:.4f}")
print(f"Log Loss with scaler:    {logloss_with_scaler:.4f}")
print(f"Improvement:             {logloss_without_scaler - logloss_with_scaler:.4f}")
print(f"Better model:            {'With Scaler' if logloss_with_scaler < logloss_without_scaler else 'Without Scaler'}")
print("\nNote: Lower log loss is better")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import log_loss
from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler
import os

glass = pd.read_csv("D:\Machine Learning\Datasets\Cases\Glass_Identification\Glass.csv")
glass
le = LabelEncoder()
glass['Type'] = le.fit_transform(glass['Type'])
X, y = glass.drop('Type', axis=1), glass['Type']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25, stratify=y)
### Without Scaling
ks = np.arange(1, 16)
k_loss_data = [] 

for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_proba = knn.predict_proba(X_test)
    loss = log_loss(y_test, y_proba) 
    k_loss_data.append([k, loss])
    
results_df = pd.DataFrame(k_loss_data, columns=['k', 'log_loss']) 
results_df.sort_values(by='log_loss', ascending=True)
### Standard Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
ks = np.arange(1, 16)
k_loss_data = [] 

for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    y_proba = knn.predict_proba(X_test_scaled)
    loss = log_loss(y_test, y_proba) 
    k_loss_data.append([k, loss])
    
results_df = pd.DataFrame(k_loss_data, columns=['k', 'log_loss']) 
results_df.sort_values(by='log_loss', ascending=True)
### MinMax Scaling
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
ks = np.arange(1, 16)
k_loss_data = [] 

for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    y_proba = knn.predict_proba(X_test_scaled)
    loss = log_loss(y_test, y_proba) 
    k_loss_data.append([k, loss])
    
results_df = pd.DataFrame(k_loss_data, columns=['k', 'log_loss']) 
results_df.sort_values(by='log_loss', ascending=True)

### Using Pipeline
mm_scaler = MinMaxScaler()
std_scaler = StandardScaler()
Ks = np.arange(1, 16)
scores = []

for k in Ks:
    for s in [std_scaler, mm_scaler, None]:
        knn = KNeighborsClassifier(n_neighbors=k)
        pipe = Pipeline(steps=[('SCI', s), ('KNN', knn)])
        pipe.fit(X_train, y_train)
        y_proba = pipe.predict_proba(X_test)
        scores.append([k, s, log_loss(y_test, y_proba)])
df_scores = pd.DataFrame(scores, columns=['k', 'scaler', 'log_loss'])
df_scores.sort_values(by='log_loss', ascending=True)

### Testing
glass_test = pd.read_csv("d:\Machine Learning\Datasets\Cases\Glass_Identification\\tst_Glass.csv")
knn = KNeighborsClassifier(n_neighbors=9)
pipe = Pipeline(steps=[('MMS', MinMaxScaler()), ('KNN', knn)])
pipe.fit(X, y)
# Apply the model to the unlabeled data
predictions = pipe.predict(glass_test)
pred_proba = pipe.predict_proba(glass_test)
glass_test['Predicted_Type'] = le.inverse_transform(predictions)
glass_test
```
