```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder,MinMaxScaler,LabelEncoder
from sklearn.naive_bayes import BernoulliNB
from sklearn.metrics import (accuracy_score, classification_report, mean_absolute_error,mean_squared_error,
                             confusion_matrix, precision_score, recall_score, 
                             f1_score, roc_auc_score, roc_curve,log_loss)
import os
import sys
from sklearn.pipeline import Pipeline
```

# Bernoulli Naive Bayes

```python
import warnings
warnings.filterwarnings('ignore')
os.chdir('D:\Machine Learning\Datasets')
cancer=pd.read_csv("d:\Machine Learning\Datasets\Cases\Cancer\Cancer.csv",index_col=1)
cancer
```

```python



le=LabelEncoder()
cancer['Class']=le.fit_transform(cancer['Class'])
```

```python
X,y=cancer.drop('Class',axis=1),cancer['Class']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25,stratify=y)
print(f"Training samples: {X_train.shape[0]}, Test samples: {X_test.shape[0]}")
```

```python
ohe=OneHotEncoder(sparse_output=False, handle_unknown='ignore').set_output(transform='pandas')
X_train_ohe=ohe.fit_transform(X_train)
X_test_ohe=ohe.transform(X_test)
```

```python
nb=BernoulliNB()
nb.fit(X_train_ohe,y_train)
y_pred_prob=nb.predict_proba(X_test_ohe)
print('log loss: ',log_loss(y_test,y_pred_prob))
y_pred=nb.predict(X_test_ohe)
print('Accuracy Score: ',accuracy_score(y_test,y_pred))
```

# HR


```python
hr=pd.read_csv("D:\Machine Learning\Datasets\HR_comma_sep.csv")
hr
```

```python
X,y=hr.drop('left',axis=1),hr['left']
```

```python
from sklearn.compose import ColumnTransformer, make_column_selector
ohe=OneHotEncoder(drop='first',sparse_output=False).set_output(transform='pandas') # one hot encoder
col_trnf=ColumnTransformer([('OHE',ohe,make_column_selector(dtype_include=object))],remainder='passthrough',verbose_feature_names_out=True)
col_trnf=col_trnf.set_output(transform='pandas')
X=col_trnf.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25,stratify=y)
print(f"Training samples: {X_train.shape[0]}, Test samples: {X_test.shape[0]}")
```

```python
nb=GaussianNB()
nb.fit(X_train,y_train)
y_pred_prob=nb.predict_proba(X_test)
print('log loss: ',log_loss(y_test,y_pred_prob))
y_pred=nb.predict(X_test)
print('Accuracy Score: ',accuracy_score(y_test,y_pred))
```
