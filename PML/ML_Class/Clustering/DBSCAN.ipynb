```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
import seaborn as sns
import os
os.chdir("D:\\Machine Learning\\Datasets")
milk = pd.read_csv("milk.csv", index_col=0)
```

```python
scaler = StandardScaler().set_output(transform="pandas")
df_scaled = scaler.fit_transform(milk)
```

```python
from sklearn.cluster import DBSCAN


clust=DBSCAN(eps=0.5,min_samples=3)
clust.fit(df_scaled)
df_scaled_cpy=df_scaled.copy()
df_scaled_cpy['label']=clust.labels_
df_scaled_cpy=df_scaled_cpy[df_scaled_cpy['label']!=1]
silhouette_score(df_scaled_cpy.iloc[:,:-1],df_scaled_cpy['label'])
if len(df_scaled_cpy['label'].unique())>=2:
    print(silhouette_score(df_scaled_cpy.iloc[:,:-1],df_scaled_cpy['label']))
else:
    print('silhouette_score cannot be cal')
```

```python
from sklearn.cluster import KMeans


km=KMeans(n_clusters=3,random_state=35)
km.fit(df_scaled)
df_cluster=km.copy()
df_cluster['cluster']=km.mean()
```
