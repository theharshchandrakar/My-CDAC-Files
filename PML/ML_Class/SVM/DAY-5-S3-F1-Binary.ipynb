# Vehicles Dataset

```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn import svm
from tqdm import tqdm
from sklearn.preprocessing import LabelEncoder,StandardScaler,OneHotEncoder

from sklearn.metrics import roc_curve, roc_auc_score, RocCurveDisplay , accuracy_score
import matplotlib.pyplot as plt

```

```python
vehicle=pd.read_csv("D:\Machine Learning\Datasets\Cases\Vehicle_Silhouettes\Vehicle.csv")
vehicle
```

```python
le = LabelEncoder()
vehicle['Class'] = le.fit_transform(vehicle['Class'])
X,y=vehicle.drop('Class',axis=1),vehicle['Class']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25,stratify=y)
print(f"Training samples: {X_train.shape[0]}, Test samples: {X_test.shape[0]}")
vehicle
```

```python
std_scaler=StandardScaler()
```

Linear

```python
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC


cs=np.linspace(.001,5,15)
dfs=['ovo','ovr']#ovo=one v one and ovr = one verses remaining
scores=[]
for s in tqdm(dfs):
    for c in cs:
        svm=SVC(kernel="linear",C=c,decision_function_shape=s)
        pipe=Pipeline([("SCL",std_scaler),("SVC",svm)])
        pipe.fit(X_train,y_train)
        y_pred=pipe.predict(X_test)
        scores.append([c,s,accuracy_score(y_test,y_pred)])
df_scores=pd.DataFrame(scores,columns=('c','decision','score'))
df_scores.sort_values('score',ascending=False).head()
```

# Glass Identification


```python
glass=pd.read_csv("D:\Machine Learning\Datasets\Cases\Glass_Identification\Glass.csv")
```

```python
le = LabelEncoder()
glass['Type'] = le.fit_transform(glass['Type'])
X,y=glass.drop('Type',axis=1),glass['Type']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25,stratify=y)
print(f"Training samples: {X_train.shape[0]}, Test samples: {X_test.shape[0]}")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC


cs=np.linspace(.001,5,15)
dfs=['ovo','ovr']#ovo=one v one and ovr = one verses remaining
kernal=['linear','poly','rbf']#kernals
scores=[]
for s in tqdm(dfs):
    for c in cs:
        for k in kernal:
            svm=SVC(kernel=k,C=c,decision_function_shape=s)
            pipe=Pipeline([("SCL",std_scaler),("SVC",svm)])
            pipe.fit(X_train,y_train)
            y_pred=pipe.predict(X_test)
            scores.append([c,s,accuracy_score(y_test,y_pred)])
df_scores=pd.DataFrame(scores,columns=('c','decision','score'))
df_scores.sort_values('score',ascending=False)
```

```python
# Get best for each decision function
best_per_decision = df_scores.loc[df_scores.groupby('decision')['score'].idxmax()]
best_per_decision
```
