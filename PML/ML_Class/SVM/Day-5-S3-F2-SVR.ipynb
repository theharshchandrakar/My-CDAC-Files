# Support Vector Regression

```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn import svm
from tqdm import tqdm
from sklearn.discriminant_analysis import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.svm import SVR
```

```python
from sklearn.metrics import roc_curve, roc_auc_score, RocCurveDisplay , accuracy_score,mean_absolute_error
```

```python
boston=pd.read_csv("D:\Machine Learning\Datasets\Boston.csv")
boston.columns
```

```python
X,y=boston.drop('medv',axis=1),boston['medv']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25)
print(f"Training samples: {X_train.shape[0]}, Test samples: {X_test.shape[0]}")
```

```python

from sklearn.svm import SVR


std_scaler=StandardScaler()

cs=np.linspace(.001,5,15)
es=np.linspace(.001,5,15)
scores=[]
for e in tqdm(es):
    for c in cs:
        svm=SVR(kernel="linear",C=c,epsilon=e)
        pipe=Pipeline([("SCL",std_scaler),("SVC",svm)])
        pipe.fit(X_train,y_train)
        y_pred=pipe.predict(X_test)
        scores.append([c,e,mean_absolute_error(y_test,y_pred)])
df_scores=pd.DataFrame(scores,columns=('c','eps','score'))
df_scores.sort_values('score',ascending=False).head()
```

```python

from sklearn.svm import SVR


std_scaler=StandardScaler()
gs=np.linspace(.001,5,15)
cs=np.linspace(.001,5,15)
es=np.linspace(.001,5,15)
scores=[]

for e in tqdm(es):
    for c in cs:
        for g in gs:
            svm=SVR(kernel="rbf",C=c,epsilon=e,gamma=g)
            pipe=Pipeline([("SCL",std_scaler),("SVC",svm)])
            pipe.fit(X_train,y_train)
            y_pred=pipe.predict(X_test)
            scores.append([c,e,g,mean_absolute_error(y_test,y_pred)])
df_scores=pd.DataFrame(scores,columns=('c','eps','gamma','score'))
df_scores.sort_values('score').head()
```

```python
from sklearn.svm import SVR


std_scaler=StandardScaler()
gs=np.linspace(.001,5,15)
cs=np.linspace(.001,5,15)
es=np.linspace(.001,5,15)
kernal=['linear','rbf','poly']
scores=[]
for k in tqdm(kernal):
    for e in es:
        for c in cs:
            for g in gs:
                svm=SVR(kernel=k,C=c,epsilon=e,gamma=g)
                pipe=Pipeline([("SCL",std_scaler),("SVC",svm)])
                pipe.fit(X_train,y_train)
                y_pred=pipe.predict(X_test)
                scores.append([c,e,g,kernal,mean_absolute_error(y_test,y_pred)])
df_scores=pd.DataFrame(scores,columns=('c','eps','gamma','kernal','score'))
df_scores.sort_values('score').head()
```
