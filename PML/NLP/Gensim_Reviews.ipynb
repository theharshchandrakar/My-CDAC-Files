```python
import gensim
gensim.__version__
import os
# os.chdir("D:\\meridianthe4\\PML\\Datasets")
import pandas as pd
from gensim.models import KeyedVectors
from gensim.models import Word2Vec
import numpy as np
import re
dataset = pd.read_csv("D:\Machine Learning\Datasets\Restaurant_Reviews.tsv", sep='\t')
dataset.head()
corpus = []
for i in range(0, dataset.shape[0]):
    review = dataset['Review'][i]
    review = re.sub('[^a-zA-Z]', ' ', review)
    review = review.lower()
    review = review.split()
    corpus.append(review)
len(corpus)
model_r = Word2Vec(corpus, min_count = 1, sg=1, vector_size=300)
means = []
for sentence in corpus:
    word_vectors = [model_r.wv.get_vector(word, norm= True) for word in sentence]
    row_means = np.mean(word_vectors,axis=0)
    means.append(row_means)
X = np.array(means)
y = dataset['Liked']
```
