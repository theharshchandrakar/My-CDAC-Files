```python
import os
os.chdir('D:\Machine Learning\Datasets')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
df = pd.read_csv('monthly-milk-production-pounds-p.csv')
df.plot()
plt.show()
```

```python
series = df['Milk']
result = seasonal_decompose(series, model='additive',period=12)
result.plot()
plt.title("Additive Decomposition")
plt.show()
```

```python
series = df['Milk']
result = seasonal_decompose(series, model='multiplicative',period=12)
result.plot()
plt.title("Multiplicative Decomposition")
plt.show()
```

### Resample to Quarterly


```python
df = pd.read_csv('monthly-milk-production-pounds-p.csv', index_col=0)
df.index = pd.to_datetime(df.index).to_period("M")
df_qtr = df.resample('Q').sum()
df_qtr.plot()
plt.title("Quarterly Total of Milk Production")
plt.show()
```

```python
df_qtr = df.resample('Q').mean()
df_qtr.plot()
plt.title("Quarterly Average of Milk Production")
plt.show()
```
