```python
import pandas as pd
from chronos import Chronos2Pipeline
from sklearn.metrics import root_mean_squared_error
import os
os.chdir("C:/Users/dbda.STUDENTSDC/Desktop/Practical Machine Learning/Datasets")
```

```python
pipeline = Chronos2Pipeline.from_pretrained("amazon/chronos-2")
```

```python
df =pd.read_csv("monthly-milk-production-pounds-p.csv")
df['timestamp'] = pd.date_range(start = '1962-01-01',periods=len(df),freq = 'ME') 
# df['timestamp'] name remain same for any dataset
df['id'] = 0  # id should be identical nos.
df.head()
```

```python
train = df.iloc[:-12]
test = df.iloc[-12:]
future_df = test.drop(columns=['Milk']).copy()
future_df['id'] = 0
future_df['timestamp'] = pd.date_range(start = '1962-01-01',periods=12,freq = 'ME')
```

```python
pred_df = pipeline.predict_df(
    train,future_df = future_df,
    validate_inputs = False,
    prediction_length = 12, # nos. of steps to forecast
    id_column = 'id', # column indentifying diff. time series
    timestamp_column ='timestamp', # column  with date time info
    target='Milk'  # column(s) with time series values to predict
)
y_pred = pred_df['predictions']
pred_df
```

```python
rmse_val = root_mean_squared_error(test['Milk'], y_pred)
print(f"RMSE: {rmse_val:.2f}")
```
